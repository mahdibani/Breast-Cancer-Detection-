#Data Preprocessing

```python
from sklearn import datasets,metrics
```

```python
cancer=datasets.load_breast_cancer()
```

```python
print(cancer.DESCR)
```

```python
# x represents the matrix of features
x=pd.DataFrame(data=cancer.data,columns=cancer.feature_names)
```

```python
x
```

```python
# dependent variable
y=cancer.target
```

```python
y
```

```python
cancer.target_names
```

# splitting the dataset in train and test set

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x,y,test_size=0.2,random_state=0)
```

```python
x_train.shape , x_test.shape
```

```python
# feature scaling
from sklearn.preprocessing import StandardScaler
sc=StandardScaler()
x_train=sc.fit_transform(x_train)
x_test=sc.transform(x_test)
```

```python
x_train
```

```python
x_train.shape , x_test.shape
```

```python
x_train=x_train.reshape(455,30,1)
x_test=x_test.reshape(114,30,1)
```

```python
x_train.shape , x_test.shape
```

# Building the Model

```python
# define an object
model = tf.keras.models.Sequential()
```

```python
# First CNN layer
input_layer = tf.keras.Input(shape=(30, 1))
conv_layer = tf.keras.layers.Conv1D(filters=32, kernel_size=2, activation='relu')(input_layer)
model = tf.keras.Model(inputs=input_layer, outputs=conv_layer)
# Batch normalisation
batch_norm_layer = tf.keras.layers.BatchNormalization()(conv_layer)
# dropout layer
dropout_layer = tf.keras.layers.Dropout(0.2)(batch_norm_layer)
model = tf.keras.Model(inputs=input_layer, outputs=dropout_layer)
```

```python
# Second CNN layer
conv_layer_2 = tf.keras.layers.Conv1D(filters=64, kernel_size=2, activation='relu')(dropout_layer)  # Use the output from the previous dropout layer
batch_norm_layer_2 = tf.keras.layers.BatchNormalization()(conv_layer_2)
dropout_layer_2 = tf.keras.layers.Dropout(0.4)(batch_norm_layer_2)

# Update the model to include the new layers
model = tf.keras.Model(inputs=input_layer, outputs=dropout_layer_2)
```

```python
# Flatten layer
flatten_layer = tf.keras.layers.Flatten()(dropout_layer_2)  # Use the output from the previous dropout layer

# Update the model to include the Flatten layer
model = tf.keras.Model(inputs=input_layer, outputs=flatten_layer)
```

```python
# Dense layer
dense_layer = tf.keras.layers.Dense(units=64, activation='relu')(flatten_layer)  # Use the output from the Flatten layer

# Output layer
output_layer = tf.keras.layers.Dense(units=1, activation='sigmoid')(dense_layer)

# Update the model to include the new layers
model = tf.keras.Model(inputs=input_layer, outputs=output_layer)
```

```python
model.summary()
```

```python
# compile the model
opt = tf.keras.optimizers.Adam(learning_rate=0.00005)
model.compile(optimizer=opt,loss='binary_crossentropy',metrics=['accuracy'])
```

# Training our Model

```python
history=model.fit(x_train,y_train,epochs=50, validation_data=(x_test,y_test))
```

# Model Prediction

```python
y_pred_probs = model.predict(x_test)
y_pred = (y_pred_probs > 0.5).astype(int)
```

```python
print(y_pred[0]),print(y_test[0])
```

```python
print(y_pred[6]),print(y_test[6])
```

```python
# confusion matrix just to check the t/f & p/n
from sklearn.metrics import confusion_matrix, accuracy_score
cm=confusion_matrix(y_test,y_pred)
print(cm)
```

```python
acc_cm = accuracy_score(y_test,y_pred)
print(acc_cm)
```

Learning Curve

```python
def learning_curve(history,epoch):
  # training vs validation accuracy
  epoch_range=range(1,epoch+1)
  plt.plot(epoch_range,history.history['accuracy'])
  plt.plot(epoch_range,history.history['val_accuracy'])
  plt.title('Model Accuracy')
  plt.ylabel('Accuracy')
  plt.xlabel('Epoch')
  plt.legend(['Train','val'],loc='upper left')
  plt.show()
  # training vs validation loss
  plt.plot(epoch_range,history.history['loss'])
  plt.plot(epoch_range,history.history['val_loss'])
  plt.title('Model Loss')
  plt.ylabel('Loss')
  plt.xlabel('Epoch')
  plt.legend(['Train','val'],loc='upper left')
  plt.show()
```

```python
learning_curve(history,50)
```
